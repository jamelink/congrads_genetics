```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os

windows = True

if windows:
    workspace_path = "P:/workspaces/lg-ukbiobank/projects/"
else:
    workspace_path = "/data/workspaces/lag/workspaces/lg-ukbiobank/projects/"

corrs = pd.read_csv(os.path.join(workspace_path, "CONGRADS_rest", 'congrads_glasser_fslcc_z_corrs_N46417.tsv'), sep='\t', index_col=0) 
```

```python
main_palette = sns.color_palette("deep", n_colors=4)
fig, ax = plt.subplots(figsize=(10,6))
sns.histplot(data=corrs.iloc[:, 1:], palette=main_palette, kde=True, bins=100, alpha=0.5, ax=ax)
ax.set_xlim(0,1)
sns.despine()
#add vertical plot at 0.3
ax.axvline(x=0.3, color='red', linestyle='--')

plt.xlabel('Spatial correlation to template')

plt.title('Distribution of spatial correlations to CONGRADS template')
plt.tight_layout()
plt.savefig(os.path.join(workspace_path, "CONGRADS_rest", 'results', 'Supp_fig_spatial_correlation_distribution_allROIs_allSubs.png'), dpi=300)
```
