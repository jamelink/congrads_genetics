```python
def count_nonzeros(fn):
    """ 
    Load data + count number of non-zeros 
    """ 
    #Load data
    print("Loading data")
    data=nib.load(fn).get_fdata()

    #count non-zeros per slice
    print("Counting non-zeros per slice")
    non_zeros = np.zeros(data.shape[3])
    for x in np.arange(data.shape[3]):
        non_zeros[x] = np.count_nonzero(data[:, :, :, x])
    del data
    
    #Do the counting
    values, counts = np.unique(non_zeros, return_counts=True)
    print("Occuring values: ", values)
    print("Counts: ", counts)
    ind = np.argmax(counts)
    most_often = values[ind]
    outlier_boolean = np.where(non_zeros != most_often)
    
    return non_zeros, values, counts, outlier_boolean
```

```python
base_path="/data/clusterfs/lag/projects/lg-ukbiobank/working_data/imaging_data/CONGRADS_rest/melodic/g1/cmap/"
roi="lifg" 
fn = os.path.join(base_path, roi, "all_data_lifg_g1_cmap_template.nii.gz")
non_zeros_cmap, values_cmap, counts_cmap, outlier_boolean_cmap = count_nonzeros(fn)
```
